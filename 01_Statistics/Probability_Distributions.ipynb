# Probability Distributions

## 📌 What is a Probability Distribution?

A Probability Distribution describes how data is spread out and how likely different values are to occur.

Instead of looking at one value, it helps us understand the overall pattern of the data.

It answers questions like:

- Where are most values?
- Which values are rare?
- Are there any unusual values?
- What does the overall shape of the data look like?

---

# Key Concepts

## 1. Cluster

A cluster is the area where most data values are grouped together.

Example:

84, 85, 86, 87, 88, 89, 90

Most values stay close together.

---

## 2. Outlier

An outlier is a value that is much farther away from the rest of the data.

Example:

21, 22, 23, 24, 25, 26, 80

Here, **80** is an outlier because it is very different from the other values.

---

# Why Probability Distributions Matter

Probability distributions help us:

- Understand how data is spread.
- Identify clusters.
- Detect outliers.
- Analyze patterns before building machine learning models.

Many AI applications use probability distributions, including:

- Spam Email Detection
- Recommendation Systems
- Medical Diagnosis
- Fraud Detection
- Weather Prediction

---

# Normal Distribution (Bell Curve)

The Normal Distribution is the most common probability distribution in statistics.

It is called the **Bell Curve** because its shape looks like a bell.

Characteristics:

- Most values are located near the center.
- Very small values are rare.
- Very large values are rare.
- The graph is approximately symmetric.

Example:

Human Heights

Most people have average heights.

Very few people are extremely short or extremely tall.

---

# Histogram

A Histogram is used to visualize the distribution of numerical data.

It shows:

- How many values fall into each range.
- Where the data is concentrated.
- Whether the data follows a Bell Curve.

Python function:

```python
plt.hist(data)
```

---

# Python Practice

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=70, scale=10, size=1000)

plt.hist(data, bins=30)
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

---

# Understanding the Parameters

## loc

The center (mean) of the distribution.

Example:

```python
loc = 70
```

Most generated values will be around **70**.

---

## scale

Controls how spread out the data is.

- Smaller scale → Narrow distribution
- Larger scale → Wider distribution

---

## size

The number of random data points generated.

Example:

```python
size = 1000
```

Generates **1000** random values.

---

# Sample Size Experiment

## size = 50

Observation:

- Few observations
- More randomness
- Histogram looks bumpy
- Bell Curve is not very clear

---

## size = 10000

Observation:

- Many observations
- Randomness averages out
- Histogram becomes smoother
- Bell Curve becomes much clearer

---

# Important Lesson

A larger sample size produces a smoother and more stable distribution because random variation averages out.

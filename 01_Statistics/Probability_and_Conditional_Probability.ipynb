📓 Probability, Independent & Dependent Events, and Conditional Probability

🎯 Learning Objectives:

In this lesson, I learned the fundamental concepts of probability and how AI uses probability to make predictions.
📌 What is Probability?
Probability measures how likely an event is to happen.
The probability of an event is always between 0 and 1.
0 → Impossible
1 → Certain

Examples:
🌞 The sun will rise tomorrow → Probability ≈ 1
🐟 A fish will drive a car → Probability = 0
AI uses probability because it cannot predict the future with 100% certainty.
Instead, it estimates how likely an event is based on available evidence.

📌 Independent Events:
Independent events do not affect each other.
Examples:
Coin flips
Dice rolls
Example:
Rolling a dice twice.
The result of the first roll does not change the probability of the second roll.

📌 Dependent Events:
Dependent events change the probability of the next event.
Examples:
Drawing cards without replacement
Picking balls from a bag without putting them back
Example:
If a red ball is removed from a bag, the probability of picking another red ball changes.

📌 Conditional Probability:
Conditional Probability means:
The probability of an event given that another event has already happened.
Keyword:
"Given"
Example:
Probability of COVID given a positive test.
New information changes the probability.

📌 Bayes' Theorem Connection:
Bayes' Theorem updates probability whenever new evidence arrives.
Example:
Before seeing the picture:
Cat = 30%
After detecting:
🐱 Pointy ears
🐱 Whiskers
🐱 Tail
The AI updates its prediction:
Cat = 97%
This is exactly how AI systems improve their predictions using new evidence.

🤖 Real AI Applications:
Spam Detection
Fraud Detection
Medical Diagnosis
Face Recognition
Image Classification
Recommendation Systems

✅ Key Takeaways:

Probability measures uncertainty.
Independent events do not affect each other.
Dependent events change future probabilities.
Conditional Probability uses new information.
Bayes' Theorem updates probabilities using new evidence.
AI makes probabilistic predictions, not perfect predictions.


🐍 Python Example 1 — Coin Flip Simulation

```python
import random

coin = random.choice(["Heads", "Tails"])
print("Result:", coin)
```

🐍 Python Example 3 — Simple Conditional Probability Example

```python
has_fever = True
positive_test = True

if has_fever and positive_test:
    print("Higher probability of disease.")
else:
    print("Need more evidence.")
```

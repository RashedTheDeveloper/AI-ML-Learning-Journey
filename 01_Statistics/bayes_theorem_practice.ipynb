📝 What is Bayes' Theorem?
Bayes' Theorem is a mathematical method for updating the probability of an event when new evidence becomes available.
It does not create information.
It simply helps us use new information wisely.


🧠 Think of it like this:

Old Belief
     │
     ▼
 New Evidence
     │
     ▼
Updated Belief

This simple flow is the heart of Bayes' Theorem.

🎯 This is the key sentence

Bayes' Theorem answers questions like: "What is the probability that A is true, given that I have observed B?"
Examples:

1. What is the probability a patient has COVID, given a positive test?
2. What is the probability an email is spam, given it contains the word "FREE"?
3. What is the probability it will rain, given dark clouds have appeared?

# Medical Test Example:

Total Population = 10,000

                Positive   Negative   Total

Sick               99          1       100

Healthy           495       9405      9900

Total             594       9406     10000

# Bayes' Formula:


P(A∣B) = P(B/A)*P(A)/P(B)

| Symbol | Meaning      |                |
| ------ | ------------ | -------------- |
| P(A)   | Prior belief |                |
| P(B    | A)           | Likelihood     |
| P(B)   | Evidence     |                |
| P(A    | B)           | Updated belief |


```python
P_A = 0.01
P_B_given_A = 0.99
P_B = 594 / 10000
```

```python
P_A_given_B = (P_B_given_A * P_A) / P_B

print(P_A_given_B)
```

```python
print(f"Probability of actually having the disease: {P_A_given_B:.2%}")
```

```python
import matplotlib.pyplot as plt

labels = ["Actually Sick", "Healthy but Positive"]
values = [99, 495]

plt.bar(labels, values)

plt.title("Positive Test Results")

plt.ylabel("People")

plt.show()
```

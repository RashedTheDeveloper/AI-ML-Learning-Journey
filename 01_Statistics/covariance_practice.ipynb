# Covariance

## Definition

Covariance measures how two variables change together.

### Interpretation

- Positive covariance → Both variables increase or decrease together.
- Negative covariance → One variable increases while the other decreases.
- Zero covariance → No linear relationship.

### Limitation

Covariance depends on the scale of the variables, so it cannot be used to compare relationships across different datasets.

```python
study_hours = [2,3,4,5,6]
exam_scores = [55,60,70,80,90]

mean_study_hours = sum(study_hours) / len(study_hours)
mean_exam_scores = sum(exam_scores) / len(exam_scores)

print(mean_study_hours)
print(mean_exam_scores)
```

```python
study_hours = [2, 3, 4, 5, 6]
exam_scores = [55, 60, 70, 80, 90]

mean_hours = sum(study_hours) / len(study_hours)
mean_scores = sum(exam_scores) / len(exam_scores)

hours_deviation = []
scores_deviation = []

for hour in study_hours:
    hours_deviation.append(hour - mean_hours)

for score in exam_scores:
    scores_deviation.append(score - mean_scores)

print("Hours Deviation:", hours_deviation)
print("Scores Deviation:", scores_deviation)
```

```python
products = []

for h_dev, s_dev in zip(hours_deviation, scores_deviation):
    products.append(h_dev * s_dev)

print(products)
```

```python
sum_products = sum(products)
print("Sum of Products:", sum_products)
```

```python
covariance = sum_products / len(study_hours)

print("Covariance:", covariance)
```

```python
# Study Hours
x = [2, 3, 4, 5, 6]

# Exam Scores
y = [95, 85, 70, 60, 50]

# Step 1: Calculate means
mean_x = sum(x) / len(x)
mean_y = sum(y) / len(y)

print("Mean of X:", mean_x)
print("Mean of Y:", mean_y)

# Step 2: Calculate covariance
covariance = sum(
    (xi - mean_x) * (yi - mean_y)
    for xi, yi in zip(x, y)
) / len(x)

print("Covariance =", covariance)
```

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5]
exam_scores = [50, 60, 70, 85, 90]

plt.scatter(study_hours, exam_scores)

plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.title("Study Hours vs Exam Scores")

plt.show()
```

# Correlation Analysis

The scatter plot shows a positive correlation between study hours and exam scores.

As study hours increase, exam scores also increase.

This means students who study more tend to achieve higher exam scores.

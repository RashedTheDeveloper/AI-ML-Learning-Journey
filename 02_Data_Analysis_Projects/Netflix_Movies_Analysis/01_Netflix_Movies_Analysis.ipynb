```python
df_countries = df.copy()

df_countries['country'] = df_countries['country'].str.split(', ')

df_countries = df_countries.explode('country')
```

```python
df_countries[['title', 'country']].head(10)
```

```python
country_counts = df_countries['country'].value_counts()

country_counts.head(10)
```

```python
real_country_counts = country_counts[country_counts.index != 'Unknown']

real_country_counts.head(10)
```

```python
top10_countries = real_country_counts.head(10)

top10_countries
```

```python
top10_countries.plot(kind='barh')

plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')

plt.tight_layout()
plt.savefig('images/top10_real_countries.png')

plt.show()
```

```python
top10_countries.sort_values().plot(kind='barh')

plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')

plt.tight_layout()
plt.savefig('images/top10_real_countries_sorted.png')
plt.show()
```

```python
ax = top10_countries.sort_values().plot(kind='barh')

plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')

for container in ax.containers:
    ax.set_xlim(0, 4000)
    
    ax.bar_label(container)

plt.tight_layout()
plt.savefig('images/top10_countries_with_values.png')
plt.show()
```

# Trend of Netflix Releases Over the Years

```python
release_counts = df['release_year'].value_counts().sort_index()

release_counts.head()
```

```python
import matplotlib.pyplot as plt

release_counts.plot(kind='line', figsize=(10,5))

plt.title("Netflix Content Released Over the Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.savefig("images/netflix_release_trend1.png")

plt.show()
```

# Top Countries Producing Netflix Content

```python
df['country'].head(10)
```

```python
df['country'].value_counts().head(10)
```

```python
country_split = df['country'].str.split(',')

country_split.head(10)
```

```python
country_exploded = country_split.explode()

country_exploded.head(20)
```

```python
df['rating'].value_counts().head(10)
```

```python
df['country'].nunique()
```

```python
df['country'].value_counts().head()
```

```python
df['rating'].nunique()
```

```python
country_counts = country_exploded.value_counts().head(10)

country_counts
```

```python
import matplotlib.pyplot as plt

country_counts.plot(kind="barh")

plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")


plt.tight_layout()
plt.savefig("images/top10_countries.png")

plt.show()
```

```python
rating_count = df['rating'].value_counts()

rating_count.plot(kind="barh")

plt.title("Distribution of Netflix Content by Rating")
plt.xlabel("Number of Titles")
plt.ylabel("Rating")

plt.tight_layout()
plt.savefig("images/rating_distribution.png")
plt.show()
```
